2 - Atendimentos com valor entre 90 e 200 reais.

```python
consulta_valor = list(colecao.find({
    "valor": {"$gte": 90, "$lte": 200}
}))

display(consulta_valor)
```

3 - Atendimentos das cidades Assis, Ourinhos, Pompeia e Marília

```python
atendimentos_cidades = list(colecao.find({
    "cidade": {"$in": ["Assis", "Ourinhos", "Pompeia", "Marília"]}
}))

display(atendimentos_cidades)
```

4 - DataFrame

```python
df = pd.DataFrame(todos_dados)
df
```

```python
df_valores = pd.DataFrame(consulta_valor)

print(f"Total de atendimentos entre R$ 90 e R$ 200: {len(df_valores)}")
display(df_valores)
```

```python
df_cidades = pd.DataFrame(atendimentos_cidades)

print(f"Total de atendimentos nestas cidades: {len(df_cidades)}")
display(df_cidades)
```

5 - Verificar valores nulos

```python
nulos = df.isnull().sum()

display(nulos)
```

6 - Calcular faturamento total

```python
faturamento_total = df['valor'].sum()

print("Faturamento Total: R$", faturamento_total)
```

7 - Faturamento por cidade.

```python
faturamento_cidade = df.groupby('cidade')['valor'].sum().sort_values(ascending=False)
faturamento_cidade
```

8 - Faturamento por serviço

```python
faturamento_servico = df.groupby('servico')['valor'].sum().sort_values(ascending=False)
faturamento_servico
```

9 - Ticket médio geral

```python
ticket_medio = df['valor'].mean()

print(f"Ticket Médio Geral: R$ {ticket_medio:.2f}")
```

10 - Ticket médio por cidade

```python
ticket_medio_cidade = df.groupby('cidade')['valor'].mean()

print("Ticket Médio Por Cidade:")
display(ticket_medio_cidade)
```

11 - Serviço mais lucrativo

```python
servico_mais_lucrativo = faturamento_servico.idxmax()
valor_mais_lucrativo = faturamento_servico.max()
print(f"Serviço mais lucrativo: {servico_mais_lucrativo} (R$ {valor_mais_lucrativo:.2f})")
```

12 - Cidade com maior faturamento.

```python
cidade_maior_faturamento = faturamento_cidade.idxmax()
valor_mais_lucrativo = faturamento_cidade.max()
print(f"Cidade com maior faturamento:{ cidade_maior_faturamento} (R$ {valor_mais_lucrativo:.2f})")
```

14 - Tres maiores atendimentos.

```python
tres_maiores_atendimentos = df.nlargest(3, 'valor')[['tutor', 'nome_pet', 'servico', 'valor', 'cidade']]
print(tres_maiores_atendimentos)
     
```
